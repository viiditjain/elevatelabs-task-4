```python
import pandas as pd
import numpy as np

from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Load Adult Income dataset
df = pd.read_csv("/content/drive/MyDrive/adult.csv")
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
categorical_features = df.select_dtypes(include=['object']).columns
numerical_features = df.select_dtypes(include=['int64', 'float64']).columns

categorical_features, numerical_features
```

```python
# Replace '?' with NaN
df.replace('?', np.nan, inplace=True)

# Fill missing categorical values with mode
for col in categorical_features:
    df[col].fillna(df[col].mode()[0], inplace=True)
```

```python
label_encoder = LabelEncoder()
df['income'] = label_encoder.fit_transform(df['income'])
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=[col for col in categorical_features if col != 'income'],
    drop_first=True
)
```

```python
scaler = StandardScaler()

df_encoded[numerical_features] = scaler.fit_transform(
    df_encoded[numerical_features]
)
```

```python
df_encoded.to_csv("adult_income_preprocessed.csv", index=False)
```
